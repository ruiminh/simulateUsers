```python
prompt1 = """

Generate ONE realistic household food-waste scenario.

The scenario must contain:

- a clearly identifiable food item at risk of being wasted

- a realistic reason why the food may be wasted

- a plausible human behavior before any intervention

The waste risk must be concrete and believable.

Avoid:

- unrealistic situations

- vague waste risk

- unnecessary storytelling

- dramatic narratives

- interventions or solutions

Output one concise paragraph only.

"""
```

```python
prompt2 = """
Generate ONE realistic household food-waste scenario.

Good example:
"A tired university student has spinach and leftover rice in the fridge that will expire tomorrow. After coming home late from class, they are likely to ignore the ingredients and order takeaway instead because they feel too exhausted to cook."

Generate a NEW scenario with similar realism.

Output one concise paragraph only.
"""
```

```python
jsonl_file = "data/scenarios.jsonl"

N = 100

# Generate and save

with open(jsonl_file, "w") as f:

    for i in tqdm(range(N)):

        response = ollama.chat(

            model="qwen2.5:3b",

            messages=[

                {

                    "role": "user",

                    "content": prompt1

                }

            ],

            options={

                "temperature": 0.8

            }

        )

        scenario_text = response["message"]["content"].strip()

        record = {

            "id": i,

            "scenario": scenario_text

        }

        f.write(json.dumps(record) + "\n")

print(f"Saved {N} scenarios to {jsonl_file}")
```
