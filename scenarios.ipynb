````python
prompt_m1 = """

You are generating ONE synthetic household food-waste scenario for research.

The goal is to generate a realistic everyday household food-waste situation.

Generate one realistic everyday household food-waste scenario.

Treat these percentages as sampling probabilities rather than strict rules.

After selecting a category, choose a realistic food item belonging to that category.

## Scenario requirements

The scenario must include:

- one clearly identifiable food item at risk of being wasted
- one realistic reason why the food may be wasted
- one plausible human behaviour before any intervention
- one concrete and believable waste risk

The situation should represent an ordinary household.

Prefer common household behaviour over unusual situations.

Avoid defaulting to the same food items, household types, or reasons for waste.

Do not disproportionately favor stereotypical examples such as salad greens, bananas, yogurt, or food being forgotten in the refrigerator.

Prefer the kinds of frequencies and combinations that would be expected in real households rather than maximizing novelty or creativity.

The generated scenario should resemble a random draw from a realistic population.

Avoid:

- interventions
- advice
- solutions
- dialogue
- dramatic events
- disasters
- unrealistic situations

Keep the scenario concise (1–3 sentences).

## Output format

Output ONLY one JSON object.

Schema

{
    "food_at_risk":"...",
    "scenario":"..."
}
"""
````

````python


TARGET = 1000
OUTPUT_FILE = "scenarios_m1_bl.jsonl"

scenarios = []

while len(scenarios) < TARGET:
    print(f"Currently have {len(scenarios)}/{TARGET}")

    response = ollama.chat(
        model="qwen2.5:3b",
        messages=[
            {
                "role": "user",
                "content": prompt_m1
            }
        ]
    )

    text = response["message"]["content"].strip()

    # Remove markdown code fences if present
    if text.startswith("```"):
        lines = text.splitlines()
        if lines[0].startswith("```"):
            lines = lines[1:]
        if lines and lines[-1].strip() == "```":
            lines = lines[:-1]
        text = "\n".join(lines)

    # Parse JSONL
    for line in text.splitlines():
        line = line.strip()
        if not line:
            continue

        try:
            obj = json.loads(line)

            # Assign a new global id
            obj["id"] = len(scenarios) + 1

            scenarios.append(obj)

            if len(scenarios) >= TARGET:
                break

        except json.JSONDecodeError:
            print("Skipping invalid JSON:")
            print(line)

    # Save progress after every batch
    with open(OUTPUT_FILE, "w", encoding="utf-8") as f:
        for obj in scenarios:
            f.write(json.dumps(obj, ensure_ascii=False) + "\n")

print(f"\nFinished! Saved {len(scenarios)} scenarios to {OUTPUT_FILE}")
````

````python
import os
TARGET = 1000

OUTPUT_FILE = "scenarios_m1_bl.jsonl"

# Resume generation if the file already exists

start_id = 1

if os.path.exists(OUTPUT_FILE):

    with open(OUTPUT_FILE, "r", encoding="utf-8") as f:

        start_id = sum(1 for _ in f) + 1

print(f"Starting from scenario {start_id}")

with open(OUTPUT_FILE, "a", encoding="utf-8") as f:

    for i in range(start_id, TARGET + 1):

        print(f"Generating {i}/{TARGET}")

        try:

            response = ollama.chat(

                model="llama3.2:latest",  # change model if needed

                messages=[

                    {

                        "role": "user",

                        "content": prompt_m1

                    }

                ],

                options={

                    "temperature": 1.0,

                }

            )

            text = response["message"]["content"].strip()

            # Remove markdown code fences

            if text.startswith("```"):

                lines = text.splitlines()

                if lines[0].startswith("```"):

                    lines = lines[1:]

                if lines and lines[-1].strip() == "```":

                    lines = lines[:-1]

                text = "\n".join(lines).strip()

            obj = json.loads(text)

            # Add global id

            obj["id"] = i

            f.write(

                json.dumps(obj, ensure_ascii=False) + "\n"

            )

            # Force write to disk

            f.flush()

            os.fsync(f.fileno())

        except json.JSONDecodeError:

            print(f"Invalid JSON at scenario {i}")

            print(text)

        except Exception as e:

            print(f"Error at scenario {i}: {e}")

print(f"Finished. Saved to {OUTPUT_FILE}")
````

````python
SEED = 42

TARGET = 1000

OUTPUT_FILE = "scenarios_m3_cat.jsonl"

MODEL = "qwen2.5:3b"
````

````python
categories = [

    "Fresh vegetables and salads",

    "Fresh fruit",

    "Bakery",

    "Drink, confectionery and desserts",

    "Other food",

    "Meat and fish",

    "Dairy and eggs",

    "Stable food",

    "Condiments, sauces, herbs and spices",

    "Canned food"

]

weights = np.array([

    29,

    6,

    13,

    2,

    8,

    4,

    4,

    5,

    1,

    1

], dtype=float)

weights /= weights.sum()
````

````python
rng = np.random.default_rng(SEED)

sampled_categories = rng.choice(

    categories,

    size=TARGET,

    replace=True,

    p=weights

)
````

````python
with open(OUTPUT_FILE, "w", encoding="utf-8") as f:

    for i, category in enumerate(sampled_categories, start=1):

        print(f"{i}/{TARGET} : {category}")

        prompt = f"""

You are generating ONE synthetic household food-waste scenario for research.

The food at risk MUST belong to this category:

{category}

Food categories are defined according to the following grouping scheme. The descriptions and examples are provided only to clarify the intended meaning of each category. They are illustrative rather than exhaustive. Food items are not limited to the examples listed below, and any food that reasonably belongs to a category may be used.

- Bakery: bread used in sandwiches or meals prepared at home; bread and cakes (packaged or unpackaged); vegetable pizza, pizza bread, and similar products.

- Canned food: canned meat and fish; canned products combining animal and vegetable ingredients; canned vegetables and fruits such as corn, beans, and pineapple; opened canned foods; brine from canned products; and similar items.

- Condiments, sauces, herbs and spices: honey, jam, olives, mayonnaise, ketchup, peanut butter, sauces, salt, sugar, herbs, spices, and related products.

- Dairy and eggs: milk, yoghurt, cheese, butter, margarine, eggs, cooked eggs, dairy products served at home, and similar foods.

- Drinks, confectionery and desserts: biscuits, chips, sweets, desserts, beer, alcoholic and non-alcoholic drinks, tea bags, coffee grounds, and related products.

- Fresh fruit: bananas, apples, melons, and other fresh fruits.

- Fresh vegetables and salads: carrots, potatoes, leafy greens, salads, and other fresh vegetables.

- Meat and fish: fresh, cooked, or prepared meat and fish; cold cuts; shellfish; seafood; and related products.

- Other food: mixed dishes prepared at home, foods containing both animal and vegetable ingredients, uncooked vegetables not covered elsewhere, and other foods that do not naturally fit into the previous categories.

- Stable food: breakfast cereals, rice, pasta, flour, grains, and similar shelf-stable products, whether uncooked or prepared at home.

Generate one realistic everyday household food-waste scenario.

Requirements:

- choose one specific food item belonging to the given category

- include one realistic reason why the food may be wasted

- include one plausible human behaviour before any intervention

- the scenario should represent an ordinary household

- no interventions

- no advice

- no dialogue

- no disasters

- no unrealistic situations

- keep the scenario concise (1–3 sentences)

Output ONLY one JSON object.

Schema:

{{

    "food_at_risk":"...",

    "food_category":"{category}",

    "scenario":"..."

}}

"""

        response = ollama.chat(

            model=MODEL,

            messages=[

                {

                    "role": "user",

                    "content": prompt

                }

            ],

            options={

                "temperature": 1.0,

                "seed": SEED + i

            }

        )

        text = response["message"]["content"].strip()

        # Remove markdown fences if present

        if text.startswith("```"):

            lines = text.splitlines()

            if lines[0].startswith("```"):

                lines = lines[1:]

            if lines and lines[-1].strip() == "```":

                lines = lines[:-1]

            text = "\n".join(lines).strip()

        try:

            obj = json.loads(text)

            obj["id"] = i

            f.write(json.dumps(obj, ensure_ascii=False) + "\n")

        except Exception as e:

            print(f"Failed at {i}: {e}")

            print(text)
````
